```python
import numpy as np
harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

Analisis: Akan muncul error jika menggunakan [5000,7000,3000] * 0.9 karena list biasa tidak dapat meembaca angka yang ada komanya

```python
import pandas as pd
data_kantin = {
'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
'harga': [12000, 4000, 10000, 4000, 8000],
'terjual': [23, 40, None, 35, 18]
}

df = pd.DataFrame(data_kantin)
print(df)
```

Kolom yang ada data kosongnya adalah kolom menu dan terjual karena terdapat tulisan none dan menurut saya jika langsung dianalsis jika belum dibersihkan, hasilnya akan kurang akurat karena tidak semua kolom memiliki nilai

```python
df = pd.read_csv('data_kantin - data_kantin.csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

kolom yang terdapat non-null yanglebih sedikit dari total baris adalah kolom menu dengan 107 non null dan kolom terjual dengan 104 non null. Artinya di kolom yang non nullnya kurang dari total baris, terdapat missing value.

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Kolom terjual diisi 0 karena data yang kosong dianggap tidak ada penjualan. Sedangkan menu dihapus jika kosong karena kita tidak tahu nama menu yang dijual, sehingga datanya kurang berguna untuk dianalisis.

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Setelah menggunakan drop_duplicates, baris yang sama akan dihapus jika tidak ada yang sama makan jumlah baris masih sama

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

Menu yang menghasilkan total_pendapatan tertinggi adalah Es Teh dengan total pendapatan sebesar 300.000,0.
