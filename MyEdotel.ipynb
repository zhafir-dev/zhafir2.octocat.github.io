```python
from abc import ABC, abstractmethod

# Class abstract
class KamarHotel(ABC):

    def __init__(self, nama_kamar, harga_dasar):
        self.__nama_kamar = nama_kamar
        self.__stok = 0
        self.__harga_dasar = harga_dasar

# Memakai method getter dan setter
    # Getter
    def get_nama_kamar(self):
        return self.__nama_kamar

    def get_stok(self):
        return self.__stok

    def get_harga_dasar(self):
        return self.__harga_dasar

    # setter stok
    def set_stok(self, jumlah):
        if jumlah < 0:
            print("Gagal, stok tidak boleh negatif.")
        else:
            self.__stok = jumlah

    # tambah stok
    def tambah_stok(self, jumlah):
        if jumlah > 0:
            self.__stok += jumlah
            print("Berhasil menambahkan stok", self.__nama_kamar + ":", jumlah, "unit.")
        else:
            print("Jumlah stok harus lebih dari 0.")

    # abstract method
    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass


# Kamar deluxe
class KamarDeluxe(KamarHotel):

    def __init__(self, nama_kamar, harga_dasar, fasilitas):
        super().__init__(nama_kamar, harga_dasar)
        self.fasilitas = fasilitas

    def tampilkan_detail(self):
        print("Deluxe :", self.get_nama_kamar(),
              "| Fasilitas:", self.fasilitas)

    def hitung_harga_total(self, jumlah_malam):
        harga = self.get_harga_dasar()
        pajak = harga * 10 / 100
        return (harga + pajak) * jumlah_malam


# Kamar standar
class KamarStandard(KamarHotel):

    def __init__(self, nama_kamar, harga_dasar, kapasitas):
        super().__init__(nama_kamar, harga_dasar)
        self.kapasitas = kapasitas

    def tampilkan_detail(self):
        print("Standar :", self.get_nama_kamar(),
              "| Kapasitas:", self.kapasitas)

    def hitung_harga_total(self, jumlah_malam):
        harga = self.get_harga_dasar()
        pajak = harga * 5 / 100
        return (harga + pajak) * jumlah_malam

# Proses transaksi
def proses_transaksi(daftar_pesanan):

    total = 0

    # Kamar Deluxe
    kamar = daftar_pesanan[0]

    kamar.tampilkan_detail()

    harga = kamar.get_harga_dasar()
    pajak = harga * 10 / 100
    total_kamar = kamar.hitung_harga_total(2)

    print("Harga Dasar/Malam: Rp", harga,
          "| Pajak(10%): Rp", pajak)

    print("Menginap: 2 malam | Subtotal: Rp", total_kamar)

    total += total_kamar

    print()

    # Kamar Standard
    kamar = daftar_pesanan[1]

    kamar.tampilkan_detail()

    harga = kamar.get_harga_dasar()
    pajak = harga * 5 / 100
    total_kamar = kamar.hitung_harga_total(1)

    print("Harga Dasar per malam: Rp", harga,
          "| Pajak(5%): Rp", pajak)

    print("Menginap: 1 malam | Subtotal: Rp", total_kamar)

    total += total_kamar

    print()
    print("TOTAL TAGIHAN: Rp", total)

# Main program
print("--- SETUP DATA KAMAR ---")

deluxe = KamarDeluxe(
    "Kamar Deluxe Sea View",
    1500000,
    "Private Pool"
)

standard = KamarStandard(
    "Kamar Standard Superior",
    500000,
    "2 Orang"
)

# Daftar pesanan
daftar_pesanan = [
    deluxe,
    standard
]

# Tambah stok
deluxe.tambah_stok(10)
standard.tambah_stok(20)


# Coba stok negatif
standard.set_stok(-5)


# Proses transaksi
print("--- STRUK PEMESANAN ---")

proses_transaksi(daftar_pesanan)
```
